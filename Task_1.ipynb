### Q3:- Take 2 numbers as input from the user.Write a program to swap the numbers without using any special python syntax.

### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
# Write your code here
# under root of (x2-x1)^2+(y2-y1)^2

# **2 is a power and **0.5 is a under root
x1=int(input('Enter x1: '))
y1=int(input('Enter y1: '))
x2=int(input('Enter x2: '))
y2=int(input('Enter y2: '))
distance=((x2-x1)**2+(y2-y1)**2)**0.5
print('Distance: ',distance)
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Write your code here
#an=a1+(n-1)*d
#an=int(input('Enter First Two Terms of Arithmetic Series: ', sep=' '))
a = int(input("Enter first term: "))
b = int(input("Enter second term: "))
n = int(input("Enter N: "))

d = b - a

nth_term = a + (n - 1) * d

print("Nth term:", nth_term)
````

### Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

````python
# Write your code here
import math

a = int(input("Numerator 1: "))
b = int(input("Denominator 1: "))
c = int(input("Numerator 2: "))
d = int(input("Denominator 2: "))

num = a*d + c*b
den = b*d

g = math.gcd(num, den)

print(num//g, "/", den//g)
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
# Write your code here
import math

H = float(input("Tank height: "))
L = float(input("Tank length: "))
B = float(input("Tank breadth: "))

h = float(input("Glass height: "))
r = float(input("Glass radius: "))

tank = H * L * B
glass = math.pi * r * r * h

print("Number of glasses:", int(tank / glass))
````
